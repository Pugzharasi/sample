````python
#Missing Values Count by Column

missing_values = df.isnull().sum()
missing_values = missing_values[missing_values > 0].sort_values(ascending=False)

plt.figure(figsize=(10, 6))

sna.barplot(
    x=missing_values.values,
    y=missing_values.index
)

plt.title("Number of Missing Values by Column")
plt.xlabel("Number of Missing Values")
plt.ylabel("Column")

plt.show()
````

````python
# Missing Values Percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100
missing_percentage = missing_percentage[missing_percentage > 0].sort_values(ascending=False)

plt.figure(figsize=(10, 6))

sna.barplot(
    x=missing_percentage.values,
    y=missing_percentage.index
)

plt.title("Percentage of Missing Values by Column")
plt.xlabel("Missing Values (%)")
plt.ylabel("Column")

plt.show()
````

## 3. ***Data Wrangling***

````python
# Check Unique Values for each variable.
for column in df.columns:
    print(f"{column}: {df[column].nunique()} unique values")
````

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
````

````python
#remove duplicate
df.drop_duplicates(inplace=True)
````

````python
#Handle missing value
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = df[col].fillna('Unknown')
    else:
        df[col] = df[col].fillna(df[col].mean())
````

````python
#check missing value
df.isnull().sum()
````

````python
#check duplicate
df.duplicated().sum()
````

````python
#Check final dataset size
df.shape
````

### What all manipulations have you done and insights you found?

We checked the dataset for missing values and handled them based on their data type. For object (text) columns, missing values were replaced with "Unknown". For numeric columns, missing values were replaced with the mean value of that column. We also checked for duplicate records and found no duplicates. These steps improved the data quality and made the dataset ready for further analysis.

````python
# Chart - 11 visualization code scatter plot Movie Duration vs Release Year
duration_data = df[
    (df['type'] == 'Movie') &
    (df['duration'].str.contains('min', na=False))
].copy()
duration_data['duration_minutes'] = (
    duration_data['duration']
    .str.replace(' min', '', regex=False)
    .astype(int)
)
plt.figure(figsize=(10, 6))
sna.scatterplot(
    data=duration_data,
    x='release_year',
    y='duration_minutes'
)
plt.title('Movie Duration vs Release Year')
plt.xlabel('Release Year')
plt.ylabel('Duration (Minutes)')
plt.show()
````

#### Chart - 14 - Correlation Heatmap

````python
# Correlation Heatmap visualization code
plt.figure(figsize=(10, 6))
numeric_data = df.select_dtypes(include='number')
correlation = numeric_data.corr()
sna.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)
plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
````

##### 1. Why did you pick the specific chart?

This chart was selected to understand the correlation between the numerical variables in the Netflix dataset.

##### 2. What is/are the insight(s) found from the chart?

The heatmap shows the strength and direction of relationships between numerical variables. Values closer to +1 indicate a strong positive relationship, while values closer to -1 indicate a strong negative relationship. Values close to 0 indicate a weak relationship.
