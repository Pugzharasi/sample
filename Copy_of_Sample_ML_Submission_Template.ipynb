## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Hypothesis 1: Movie duration differs across different release-year groups.   
Hypothesis 2: Content type (Movie/TV Show) is related to content rating.    
Hypothesis 3: The average release year differs across different rating groups.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Research Hypothesis: Movie duration differs across different release-year groups.

Null Hypothesis (H₀): There is no significant difference in movie duration across different release-year groups.

Alternate Hypothesis (H₁): There is a significant difference in movie duration across different release-year groups.

#### 2. Perform an appropriate statistical test.

````python
from scipy.stats import f_oneway

# Select only movies
movies = df[df['type'] == 'Movie'].copy()

# Convert movie duration into numeric values
movies['duration_num'] = movies['duration'].str.replace(' min', '', regex=False).astype(float)

# Create release-year groups
movies['year_group'] = pd.cut(
    movies['release_year'],
    bins=[1900, 2000, 2010, 2020, 2030],
    labels=['Before 2000', '2000-2010', '2011-2020', 'After 2020']
)

# Create groups for ANOVA
duration_groups = [
    group['duration_num'].dropna()
    for _, group in movies.groupby('year_group', observed=False)
    if len(group['duration_num'].dropna()) > 1
]

# Perform ANOVA
f_stat, p_value = f_oneway(*duration_groups)

print("F-statistic:", f_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
````

##### Which statistical test have you done to obtain P-Value?

I used the One-Way ANOVA (Analysis of Variance) test to obtain the P-value.

##### Why did you choose the specific statistical test?

One-Way ANOVA was chosen because movie duration is a numerical variable and we are comparing its average across more than two release-year groups. The test helps determine whether there is a statistically significant difference in movie duration between the groups.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Research Hypothesis: Content type (Movie/TV Show) is related to content rating.

Null Hypothesis (H₀): There is no significant relationship between content type and content rating.

Alternate Hypothesis (H₁): There is a significant relationship between content type and content rating.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
table = pd.crosstab(df['type'], df['rating'])
chi2, p_value, dof, expected = chi2_contingency(table)
print("Chi-square statistic:", chi2)
print("P-value:", p_value)
if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
````

##### Which statistical test have you done to obtain P-Value?

Chi-Square Test of Independence was used to obtain the P-value.

##### Why did you choose the specific statistical test?

The Chi-Square Test of Independence was chosen because both content type (Movie/TV Show) and content rating are categorical variables. The test helps determine whether there is a significant relationship between content type and content rating.

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Research Hypothesis: The average release year differs across different rating groups.

Null Hypothesis (H₀): There is no significant difference in the average release year across different rating groups.

Alternate Hypothesis (H₁): There is a significant difference in the average release year across different rating groups.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
top_ratings = df['rating'].value_counts().head(10).index

# Create release year groups for each rating
rating_groups = [
    df.loc[df['rating'] == r, 'release_year'].dropna()
    for r in top_ratings
]

# Perform One-Way ANOVA test
f_stat, p_value = f_oneway(*rating_groups)

print("F-statistic:", f_stat)
print("P-value:", p_value)

if p_value < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
````

##### Which statistical test have you done to obtain P-Value?

One-Way ANOVA (Analysis of Variance) test was used to obtain the P-value.

##### Why did you choose the specific statistical test?

One-Way ANOVA was chosen because release year is a numerical variable, and we are comparing the average release year across multiple rating groups. The test helps determine whether there is a statistically significant difference in the average release year between the rating groups.

Run this code in your Google Colab notebook to evaluate the model and display a score chart.

````python
# Visualizing evaluation Metric Score chart
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import matplotlib.pyplot as plt

# Predict on test data
y_pred = model.predict(X_test)

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted', zero_division=0)
recall = recall_score(y_test, y_pred, average='weighted', zero_division=0)
f1 = f1_score(y_test, y_pred, average='weighted', zero_division=0)

# Display scores
print("Accuracy:", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
print("F1 Score:", round(f1, 4))

# Plot evaluation metric score chart
metrics = ['Accuracy', 'Precision', 'Recall', 'F1 Score']
scores = [accuracy, precision, recall, f1]

plt.figure(figsize=(8, 5))
plt.bar(metrics, scores)
plt.ylim(0, 1)
plt.title('Random Forest Classifier - Evaluation Metrics')
plt.ylabel('Score')
plt.xlabel('Evaluation Metrics')

for i, score in enumerate(scores):
    plt.text(i, score + 0.02, f'{score:.2f}', ha='center')

plt.show()
````

C. Explanation of Evaluation Metrics

- Recall: Measures how well the model identifies the actual instances of each class.

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# ML Model - 1 Implementation with hyperparameter optimization techniques (i.e., GridSearch CV, RandomSearch CV, Bayesian Optimization etc.)
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
from imblearn.pipeline import Pipeline
from imblearn.over_sampling import RandomOverSampler

# Create a pipeline
pipeline = Pipeline([
    ('sampling', RandomOverSampler(random_state=42)),
    ('classifier', RandomForestClassifier(random_state=42))
])

# Define hyperparameters
param_grid = {
    'classifier__n_estimators': [100, 200],
    'classifier__max_depth': [None, 10, 20],
    'classifier__min_samples_split': [2, 5]
}

# Apply GridSearchCV
grid_search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    cv=3,
    scoring='f1_weighted',
    n_jobs=-1
)
# Fit the Algorithm
grid_search.fit(X_train, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best Cross-Validation F1 Score:", grid_search.best_score_)
# Predict on the model
# Use the best model to predict test data
best_model = grid_search.best_estimator_
y_pred = best_model.predict(X_test)

print("Predicted values:")
print(y_pred)
````
